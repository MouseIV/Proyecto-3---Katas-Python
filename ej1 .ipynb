```python
def frecuencia_letras(cadena):
    frecuencias = {}
    for letra in cadena:
        if letra != " ":
            letra = letra.lower()  # opcional: ignorar mayúsculas
            frecuencias[letra] = frecuencias.get(letra, 0) + 1
    return frecuencias

# Ejemplo
print(frecuencia_letras("Hola Mundo"))
```
